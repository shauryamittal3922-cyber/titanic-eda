```python
import pandas as pd

df = pd.read_csv('train.csv')
df.head()
```

```python
print("Dataset Shape (Row, Column): ", df.shape)
print("\n--- Column Data Types& Non-Null Counts ---")
df.info()

print("\n--- Summary Stats for Numbers ---")
df.describe()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df.drop(columns = ['Cabin'], inplace = True, errors = 'ignore')

print("Remaining missing values: ")
print(df.isnull().sum())
df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style = "whitegrid")

plt.figure(figsize = (6, 4))
sns.countplot(x = 'Survived', data = df, palette = 'Set2')
plt.title('Count of Survivors (0 = No, 1 = Yes)')
plt.show()

plt.figure(figsize = (6, 4))
sns.countplot(x = 'Pclass', data = df, palette = 'Set2')
plt.title('Passenger Class Distribution (1st, 2nd, 3rd)')
plt.show()

plt.figure(figsize = (8, 4))
sns.histplot(df['Age'], kde = True, bins = 30, color = 'blue')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize = (6, 4))
sns.barplot(x = 'Sex', y = 'Survived', data = df, palette = 'Set3')
plt.title('Survival Rate by Gender')
plt.ylabel('Survival probability')
plt.show()

plt.figure(figsize = (6, 4))
sns.barplot(x = 'Pclass', y = 'Survived', data = df, palette = 'Set3')
plt.title('Survival Rate by Class')
plt.ylabel('Survival Probability')
plt.show()

plt.figure(figsize = (10, 5))
sns.histplot(x = 'Age', hue = 'Survived', data = df, kde = True, bins = 30, palette = 'Set2')
plt.title('Age Distribution by Survival Status (0 = Perished, 1 = Survived)')
plt.show()
```

```python
import numpy as np
numeric_df = df.select_dtypes(include = [np.number])

corr_matrix = numeric_df.corr()

plt.figure(figsize = (8, 6))
sns.heatmap(corr_matrix, annot = True, cmap = 'coolwarm', fmt = '.2f', linewidths = 0.5)
plt.title('Correlation Heatmap of Titanic Numerical Features')
plt.show()
```
